```python
from transformers import pipeline

print("Loading the AI sentiment analysis brain...")
classifier = pipeline("sentiment-analysis")
print("Model loaded successfully!\n")


test_reviews = [
    "I absolutely love this new keyboard, the tactile switches feel amazing!",
    "The shipping took three weeks and the box arrived damaged. Terrible service.",
    "The movie wasn't bad, but it definitely wasn't the masterpiece everyone claimed.",
    "I am completely indifferent about this purchase."
]


print("--- Running Sentiment Analysis ---")
for review in test_reviews:

    raw_result = classifier(review)
    prediction = raw_result[0]

    label = prediction['label']
    confidence_score = prediction['score']


    percentage = f"{confidence_score * 100:.1f}%"


    print(f"\nText: \"{review}\"")
    print(f"Result: {label} ({percentage} confidence)")
```
